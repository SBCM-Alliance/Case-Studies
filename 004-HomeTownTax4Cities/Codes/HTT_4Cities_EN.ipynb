### Generate English versions of all graphs

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ==========================================
# 2. データの定義（OCRデータより転記）
# ==========================================
# 単位：円

data = {
    "年度": ["R1", "R2", "R3", "R4", "R5", "R6"],
    # 勝浦市
    "勝浦_受入額": [1127204121, 963121879, 2356052000, 5534178000, 5145389300, 1665885900],
    "勝浦_経費":   [512198777,  524436821, 1274115990, 2764237984, 2551597239, 824090499],
    # 浦安市
    "浦安_受入額": [30941050,   39265000,  271136940,  569979000,  1195235353, 1341196798],
    "浦安_経費":   [9481496,    16481093,  119456263,  264284005,  580385541,  663030859],
    # 市川市
    "市川_受入額": [40864553,   48128302,  126516910,  55152657,   32067530,   82088616],
    "市川_経費":   [4638192,    5003249,   10559266,   8946042,    13450382,   27432324],
    # 南房総市
    "南房総_受入額": [711079795, 411461841, 388218371, 438459679, 416443700, 594520000],
    "南房総_経費":   [300760290, 11638200,  183531758, 219179875, 207571393, 292509500]
}

df = pd.DataFrame(data)

# 単位を「億円」に変換
cols = [c for c in df.columns if "受入額" in c or "経費" in c]
for c in cols:
    df[c + "_億円"] = df[c] / 100000000


# Temporarily disable Japanize-matplotlib font settings for English plots if needed, or explicitly set a common font.
# However, typically, English characters display fine with default matplotlib settings if not explicitly forced to a Japanese font.
# Let's override font settings specifically for these plots to ensure English characters are rendered correctly.
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial'] # Common sans-serif fonts

# --- Graph 1 (English): Time series of acceptance amount ---
plt.figure(figsize=(12, 7))
plt.plot(df["年度"], df["勝浦_受入額_億円"], marker='o', linewidth=3, label='Katsuura City', color='#d62728')
plt.plot(df["年度"], df["浦安_受入額_億円"], marker='s', linewidth=3, label='Urayasu City', color='#1f77b4')
plt.plot(df["年度"], df["南房総_受入額_億円"], marker='^', linewidth=3, label='Minamiboso City', color='#2ca02c')
plt.plot(df["年度"], df["市川_受入額_億円"], marker='x', linewidth=3, label='Ichikawa City', color='#9467bd')

plt.title("Furusato Nouzei Acceptance Amount Trend for 4 Cities in Chiba Prefecture (FY R1-R6)", fontsize=18, fontweight='bold')
plt.ylabel("Acceptance Amount (Billion JPY)", fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)

# Display values labels
for i, txt in enumerate(df["勝浦_受入額_億円"]):
    plt.annotate(f"{txt:.1f}", (df["年度"][i], df["勝浦_受入額_億円"][i]), xytext=(0,10), textcoords='offset points', ha='center', fontsize=10)

plt.savefig('Furusato_Nouzei_Acceptance_Trend_English.png')
plt.close() # Close to free memory


# --- Graph 2 (English): Katsuura City FY R4 (peak) cost structure ---
katsuura_r4_breakdown_en = {
    "Return Gift Procurement Costs": 16.60,
    "Shipping Costs": 6.86,
    "Payment Fees, etc.": 1.60 + 0.52,
    "Administrative Costs": 1.35,
    "[Profit] Municipal Revenue": 55.34 - 27.64
}

labels_en = katsuura_r4_breakdown_en.keys()
sizes_en = katsuura_r4_breakdown_en.values()
colors = ['#ff9999', '#ffcc99', '#99ff99', '#ffb3e6', '#c2c2f0']
explode = (0.05, 0.05, 0.05, 0.05, 0.1)

plt.figure(figsize=(10, 8))
plt.pie(sizes_en, explode=explode, labels=labels_en, autopct='%1.1f%%',
        shadow=True, startangle=90, colors=colors, textprops={'fontsize': 14})
plt.title("Breakdown of 'Lost Taxes' in Katsuura City FY R4 (Peak Period)\nTotal Acceptance Amount: Approx. 55.3 Billion JPY", fontsize=18, fontweight='bold')
plt.axis('equal')
plt.savefig('Katsuura_R4_Expense_Breakdown_English.png')
plt.close() # Close to free memory


# --- Graph 3 (English): FY R5 Comparison of 4 Cities' Net Revenue and Expense Ratio ---
cities_en = ['Katsuura City', 'Urayasu City', 'Minamiboso City', 'Ichikawa City']
income_r5 = [df.loc[4, "勝浦_受入額_億円"], df.loc[4, "浦安_受入額_億円"], df.loc[4, "南房総_受入額_億円"], df.loc[4, "市川_受入額_億円"]]
expense_r5 = [df.loc[4, "勝浦_経費_億円"], df.loc[4, "浦安_経費_億円"], df.loc[4, "南房総_経費_億円"], df.loc[4, "市川_経費_億円"]]
net_r5 = [i - e for i, e in zip(income_r5, expense_r5)]
ratios = [e/i*100 for i, e in zip(income_r5, expense_r5)]

x = np.arange(len(cities_en))
width = 0.35

fig, ax1 = plt.subplots(figsize=(12, 7))

# Stacked bar chart (Expenses + Net Revenue)
p1 = ax1.bar(x, net_r5, width, label='Municipal Net Revenue', color='#3498db')
p2 = ax1.bar(x, expense_r5, width, bottom=net_r5, label='Lost Expenses (Return Gifts, Shipping, etc.)', color='#e74c3c')

# Line graph (Expense Ratio)
ax2 = ax1.twinx()
ax2.plot(x, ratios, color='black', marker='o', linewidth=2, linestyle='--', label='Expense Ratio (%)')
ax2.set_ylim(0, 100)
ax2.set_ylabel('Expense Ratio (%)', fontsize=14)

ax1.set_ylabel('Amount (Billion JPY)', fontsize=14)
ax1.set_title('Comparison of Revenue Structure and Expense Ratio for 4 Cities in FY R5', fontsize=18, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(cities_en, fontsize=14)
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')

# Display values
for i in range(len(cities_en)):
    ax1.text(x[i], net_r5[i]/2, f"{net_r5[i]:.1f}B JPY", ha='center', color='white', fontweight='bold')
    ax1.text(x[i], net_r5[i] + expense_r5[i]/2, f"{expense_r5[i]:.1f}B JPY", ha='center', color='white', fontweight='bold')
    ax2.text(x[i], ratios[i]+3, f"{ratios[i]:.1f}%", ha='center', color='black', fontweight='bold')

plt.grid(False)
plt.savefig('R5_4Cities_Revenue_Expense_Comparison_English.png')
plt.close() # Close to free memory

# Reset font settings to original if desired for subsequent Japanese plots
# plt.rcParams['font.family'] = 'IPAexGothic' # Or whatever the original setting was
```
